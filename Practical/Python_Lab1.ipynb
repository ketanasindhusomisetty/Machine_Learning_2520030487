```python
# ---ARITHMETIC OPERATORS ---
a = 7
b = 2
print("a + b", a + b)
print("a - b", a - b)
print("a * b", a * b)
print("a / b", a / b)
print("a % b", a % b)
print("a ** 2", a ** 2)
print("a // b", a // b)
```

```python
# -- f string --
model_name = "Logistic Regression"
accuracy = 0.13408

# Old, clunky way (avoid this):
print("Accuracy of " + model_name + " is " + str(accuracy))

# f-string way (use this):
print(f"Accuracy of {model_name} is {accuracy}")

# ':.3f' inside the braces rounds a float to 3 decimal places -
# you will use this in almost every experiment to print clean metrics.
print(f"Accuracy of {model_name} is {accuracy:.3f}")
```

```python
# --- COMPARISION OPERATIONS ---
cgpa = 8.4
print("cgpa > 8 :", cgpa > 8)
print("cgpa < 8 :", cgpa < 8)
print("cgpa >= 8 :", cgpa >= 8)
print("cgpa < 8 :", cgpa <= 8)
print("cgpa == 8 :", cgpa == 8)
print("cgpa != 8 :", cgpa != 8)
```

```python
# --- LOGICAL OPERATORS: and / or / not ---
attendance = 85
placed = (cgpa > 85) and (attendance > 80)
print("placed:", placed)
placed = (cgpa > 85) or (attendance > 80)
print("placed:", placed)
placed = not (cgpa > 85)
print("placed:", placed)
```

```python
# Create a dictionary
student = {
    "Name": "Sindhu",
    "Roll No": 487,
    "CGPA": 9.0
}

# Display all key-value pairs
print("Student Details:")
for key, value in student.items():
    print(key, ":", value)

# Access a specific key
print("\nStudent Name:", student["Name"])

# Machine Learning models dictionary
ml_models = {
    "Linear Regression": "Regression",
    "Decision Tree": "Classification",
    "Random Forest": "Ensemble"
}

print("\nMachine Learning Models:")
for model, model_type in ml_models.items():
    print(model, ":", model_type)
```

```python
# Program 6: Conditional Statements

# 1. Classify student's CGPA
cgpa = 8.3

if cgpa >= 8.5:
    print("Category: High")
elif cgpa >= 6.5:
    print("Category: Mid")
else:
    print("Category: Low")

# 2. Check for missing values in a dataset
data = [10, 20, None, 40]

if None in data:
    print("Missing values are present.")
else:
    print("No missing values.")

# 3. Check placement eligibility
if cgpa >= 7.0:
    print("Student is Eligible for Placement.")
else:
    print("Student is Not Eligible for Placement.")

# 4. Assign grades based on marks
marks = int(input("Enter Marks: "))

if marks >= 90:
    print("Grade: A")
elif marks >= 75:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 40:
    print("Grade: D")
else:
    print("Grade: F")
```
